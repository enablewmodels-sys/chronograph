# EEG recording without hardware

A deterministic synthetic acquisition and durable offline spool. No server, credentials or physical device required. Install `sdk/python[bci]` from this checkout first. Synthetic data is not a clinical or device qualification.

```python
from pathlib import Path
import tempfile
import numpy as np
from chronograph_connectors.bci import Session
from chronograph_connectors.bci_spool import BCISpool
from chronograph_connectors.bci_acquisition import synthetic
root = tempfile.TemporaryDirectory(prefix="chronodb-bci-notebook-")
queue = BCISpool(Path(root.name)/"recording", "bci_research", "notebook")
session = Session(queue, clock_domain="simulation_us", participant="synthetic")
synthetic(session, seconds=8, channels=8, rate=250, seed=42)
assert queue.status()["pending_batches"] > 30
queue.status()
```

The queue retains exact signal tensors and timestamps while offline. Upload with `queue.drain(client)` only after creating a matching `simulation_us` binding in a disposable project. Use the complete `research_pipeline.py` example for three independent recordings and actual decoder training.

```python
queue.close()
root.cleanup()
```
